Section D — AI-Augmented Learning
1. The Exact Prompt Given to the AI:
"Write a Python OpenCV program that reads a food image, resizes it to 300x300, converts to grayscale, applies a 5x5 Gaussian blur, and runs Canny edge detection. Count and print the total edge pixels using cv2.countNonZero(). Then, display the original colour image and the edge map side-by-side in a single window using cv2.hconcat(). Finally, save the edge map as ai_edge_output.jpg and print a confirmation message."

2. The Code Files:
See ai_original.py and ai_corrected.py located in this folder.

3. Explanation of Changes:
The AI's code immediately crashed with a shape mismatch error on cv2.hconcat. I had to make three corrections:

I added a file-not-found check so it doesn't crash if the image is missing.
I used the resized color image for concatenation instead of the original, so the heights perfectly match (300px).
I converted the 1-channel grayscale edge map into a 3-channel image (cv2.COLOR_GRAY2BGR) so it could be successfully concatenated with the 3-channel color image without throwing an array depth error.
