```python
import cv2
import numpy as np
import sys

image_path = 'food_sample.jpg'
img = cv2.imread("C:\\Users\\Aaryan\\data science\\Images\\grayscale_food.jpg")

if img is None:
    print("Error: Image not found.")
    sys.exit()

resized_img = cv2.resize(img, (256, 256), interpolation=cv2.INTER_AREA)

b, g, r = cv2.split(resized_img)

b_mean = np.mean(b)
g_mean = np.mean(g)
r_mean = np.mean(r)

print(f"Average Pixel Intensity -> Blue: {b_mean:.2f}, Green: {g_mean:.2f}, Red: {r_mean:.2f}")

highest_channel = max(('Blue', b_mean), ('Green', g_mean), ('Red', r_mean), key=lambda item: item[1])
print(f"Summary: The {highest_channel[0]} channel has the highest intensity, suggesting it is the dominant colour in the dish.")

cv2.imshow('Resized Original', resized_img)
cv2.imshow('Blue Channel', b)
cv2.imshow('Green Channel', g)
cv2.imshow('Red Channel', r)

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Task-3

```python
import cv2
import sys

image_path = 'label_sample.jpg'
img = cv2.imread("C:\\Users\\Aaryan\\data science\\Images\\label_sample.jpg")

if img is None:
    print("Error: Image not found.")
    sys.exit()

height, width = img.shape[:2]
center = (width // 2, height // 2)

matrix = cv2.getRotationMatrix2D(center, -90, 1.0)
rotated_img = cv2.warpAffine(img, matrix, (width, height))

r_height, r_width = rotated_img.shape[:2]
cropped_img = rotated_img[r_height//2:r_height, r_width//2:r_width]

flipped_img = cv2.flip(cropped_img, 1)

cv2.imshow('1 - Rotated 90 Degrees Clockwise', rotated_img)
cv2.imshow('2 - Cropped Lower-Right Quadrant', cropped_img)
cv2.imshow('3 - Flipped Horizontally', flipped_img)

blurred_img = cv2.GaussianBlur(flipped_img, (5, 5), 0)

edges = cv2.Canny(blurred_img, 100, 200)

cv2.imshow('Label Edges', edges)

cv2.imwrite('label_edges.jpg', edges)
print("Saved edge map successfully as 'label_edges.jpg'")

cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Task-4

```python
import cv2
import sys

cascade_path = cv2.data.haarcascades + 'haarcascade_frontalface_default.xml'
face_cascade = cv2.CascadeClassifier(cascade_path)
if face_cascade.empty():
    print("Error: Failed to load cascade classifier. Reinstall opencv-python.")
    sys.exit()

cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Error: Could not open webcam.")
    sys.exit()

print("Press 's' to save a snapshot, 'q' to quit.")

while True:
    ret, frame = cap.read()
    if not ret:
        break
        
    gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    
    faces = face_cascade.detectMultiScale(gray, scaleFactor=1.1, minNeighbors=6, minSize=(80, 80))
    
    for (x, y, w, h) in faces:
        cv2.rectangle(frame, (x, y), (x+w, y+h), (0, 255, 0), 2)
        cv2.putText(frame, 'Agent Detected', (x, y - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2)
        
    cv2.imshow('Delivery Agent Verifier', frame)
    
    key = cv2.waitKey(1) & 0xFF
    if key == ord('s'):
        cv2.imwrite('agent_snapshot.jpg', frame)
        print("Snapshot saved successfully as 'agent_snapshot.jpg'")
    elif key == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```
