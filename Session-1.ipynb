# Task 1: TensorFlow Playground Experiment

## Objective
Create a neural network with:
- 1 Hidden Layer
- 3 Neurons
- Classify Orange and Blue dots

### Steps Performed

1. Opened https://playground.tensorflow.org/
2. Selected the default classification dataset.
3. Added:
   - Hidden Layers = 1
   - Neurons = 3
4. Used Tanh activation function.
5. Trained the model using different learning rates.

---

## Observation 1: Low Learning Rate (0.001)

### Result
- Training was very slow.
- Decision boundary changed gradually.
- Required many epochs to converge.
- Stable learning with fewer fluctuations.

### Advantages
- Smooth convergence
- Less chance of overshooting
- More stable training

### Disadvantages
- Takes longer to train

---

## Observation 2: Medium Learning Rate (0.03)

### Result
- Faster convergence.
- Decision boundary formed quickly.
- Model reached high accuracy.
- Stable optimization.

### Advantages
- Balanced speed
- Good accuracy
- Recommended learning rate

---

## Observation 3: High Learning Rate (1)

### Result
- Training became unstable.
- Decision boundary kept changing.
- Loss oscillated.
- Sometimes failed to converge.

### Advantages
- Very fast updates

### Disadvantages
- Overshoots minimum loss
- Poor convergence
- Unstable model

---

## Conclusion

| Learning Rate | Training Speed | Stability | Performance |
|---------------|---------------|-----------|-------------|
| 0.001 | Slow | High | Good |
| 0.03 | Moderate | High | Best |
| 1.0 | Very Fast | Low | Poor |

**Best Learning Rate:** 0.03

# Task 2: Biological Neuron vs Artificial Neuron

## Comparison Diagram

```

Biological Neuron

Dendrites
│
▼
Receive Signals
│
▼
Soma (Cell Body)
│
Processes Information
│
▼
Axon
│
▼
Output Signal

```

```

Artificial Neuron

Input Features (x1, x2, x3)
│
▼
Weights (w1, w2, w3)
│
▼
Weighted Sum
Σ (x × w) + Bias
│
▼
Activation Function
(ReLU / Sigmoid / Tanh)
│
▼
Output

```

---

## Comparison Table

| Biological Neuron | Artificial Neuron |
|-------------------|-------------------|
| Dendrites receive signals | Inputs receive data |
| Soma processes information | Weighted sum performs computation |
| Synapses determine signal strength | Weights determine importance |
| Axon sends output | Output node gives prediction |
| Brain learns from experience | Neural network learns from data |
